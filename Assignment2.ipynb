```python
#P1: A function that checks if the length of a string within a list is greater than k and returns it
def check_str(l,k):
    new = []
    for i in l:
        if len(i)>k:
            new.append(i)
    return new
check_str(["apple", "pie", "banana"],3)   
```

```python
#P2: A function that calculates the area of a circle or a rectangle
def area(shape,r_l,w=0):
    if shape.strip().lower()=='circle':
        pi = 3.14159
        print(f"Area of circle:{pi*(r_l**2)}")
    elif shape.strip().lower()=='rectangle':
        print(f"Area of rectangle:{r_l*w}")
area('rectangle',3,5)        
```

```python
#P4: Summing numbers within a list and ignoring strings or booleans
start = True
total = []
while start==True:
    state = ''
    e = input("Enter an element to add it to a list: ").strip()
    if e.isdigit()==True:
        e = int(e)
        total.append(e)
    while state=='':    
        state = input("Add more?? Enter yes or no only!!: ").strip().lower() 
        if state == 'yes':
            break
        elif state == 'no':
            start = False
            break
        else:
            state = ''
print(f"Total Sum: {sum(total)}")
```

```python
# P5: Making a restaurant class with some operations 
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = {}
        self.customer_orders = []

    
    def add_item_to_menu(self, item, price):
        if item in self.menu_items:
            print(f"{item} is already in the menu")
        else:
            self.menu_items[item] = price
            print(f"Added {item} to the menu for {price}.")
 
    def book_tables(self, table_num, name):
        if table_num in self.book_table:
            print("Sorry, this table is reserved!")
        else:
            self.book_table[table_num] = name
            print("Reservation successfully made!")

    def customer_order(self, table, orders):
        # Check if the table number already exists in any of the active orders
        self.customer_orders.append((table,orders))
        print(f"Order recorded for Table {table}: {orders}")
        

    def display_info(self):
        print(f"The menu: {self.menu_items}")
        print(f"The reservations: {self.book_table}")
        print(f"The customer orders: {self.customer_orders}")
        
# Testing the implementation
Fish_R = Restaurant()
Fish_R.add_item_to_menu('shrimps', '$40')
Fish_R.add_item_to_menu('fish', '$30')
Fish_R.add_item_to_menu('lobster', '$50')
Fish_R.add_item_to_menu('shrimps', '$40')
Fish_R.book_tables(12, 'Sendibad')
Fish_R.customer_order(12, ['shrimps', 'lobster', 'fish'])
Fish_R.display_info()
```

```python
# P6: A bike rental system
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age 

    def get_name(self):
        return self.name

    def set_name(self, name):
        if isinstance(name,str):
            self.name = name
        else:
            self.name = ""

    def get_age(self):
        return self.age

    def set_age(self, age):
        if age>6:
            self.age = age
        else:    
            print("Too young")

class Bike:

    def __init__(self, name, price):
        self.set_name(name)
        self.set_price(price) 

    def get_name(self):
        return self.name

    def set_name(self, name):
        if isinstance(name,str):
            self.name = name
        else:
            self.name = ""

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.Rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes_list):
        self.bikes = bikes_list

    def get_rental(self):
        return self.Rentals

    def set_rental(self, rentals_list):
        self.Rentals = rentals_list

    def addBike(self, bike):
        self.bikes.append(bike)

    def totalCost(self):
        total = 0
        for bike in self.bikes:
            total += bike.get_price()
        return total

    def requestBike(self,customer,bikes_num):
        if customer.get_age() > 6:
            if bikes_num <= len(self.bikes):
                del self.bikes[-bikes_num:]
                
                self.Rentals.append((customer.get_name(), bikes_num))
                print(f"Successfully rented {bikes_num} bikes!")
                return True
            else:
                print("wait 10 mins.")
                return False
        else:
            print("Rental denied: Age must be over 6 years.")
            return False
        
    def returnBike(self, rentalTime, num_of_bikes):
        total_bill = rentalTime * 40
        print(f"Total bill cost: ${total_bill}")

        for i in range(num_of_bikes):
            self.addBike(Bike("Returned Bike", 40))
            
        return total_bill



shop = MainBikeRental()

shop.addBike(Bike("Mountain Bike", 500))
shop.addBike(Bike("Road Bike", 700))
shop.addBike(Bike("City Cruiser", 300))

print(f"Starting stock count: {len(shop.get_bikes())}")

adult = Customer("Sarah", 28)
child = Customer("Leo", 5)

shop.requestBike(child, 1)  
shop.requestBike(adult, 2)  

print(f"Stock count after rental: {len(shop.get_bikes())}")

shop.returnBike(3, 2)        

print(f"Final stock count: {len(shop.get_bikes())}")
print(f"Final rentals log record: {shop.get_rental()}")
```
